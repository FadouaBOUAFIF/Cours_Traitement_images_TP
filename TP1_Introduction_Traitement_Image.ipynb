### 2.3. Images Couleurs (RGB)
La perception humaine des couleurs repose sur les cônes sensibles au Rouge (R), au Vert (G) et au Bleu (B). Sous Python, une image couleur RGB est représentée par un **tenseur 3D** de dimensions `(Hauteur, Largeur, 3)`.

```python
R = np.array([[0.5, 0.0, 1.0],
              [0.0, 1.0, 0.0],
              [1.0, 0.0, 0.5]])

G = np.array([[0.5, 0.6, 1.0],
              [0.6, 0.0, 0.0],
              [0.0, 0.6, 0.5]])

B = np.array([[0.5, 1.0, 0.0],
              [1.0, 0.0, 1.0],
              [0.0, 1.0, 0.5]])

# Empilement des 3 matrices selon la 3ème dimension (profondeur)
RGB = np.dstack((R, G, B))

print("Dimensions du tenseur RGB :", RGB.shape)

plt.figure(figsize=(4, 4))
plt.imshow(RGB)
plt.title("Image Synthèse RGB")
plt.show()
```
